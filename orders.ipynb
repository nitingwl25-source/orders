```python
import pandas as pd
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("ankitbansal06/retail-orders")

print("Path to dataset files:", path)
```

```python
import os
```

```python
file = os.listdir(path)[0]

df = pd.read_csv(os.path.join(path, file))

df.head()
```

```python
df.shape
```

```python
df.head(5)
```

```python
df["Ship Mode"].value_counts()
```

```python
import numpy as np
```

```python
df["Ship Mode"]= df["Ship Mode"].replace({
    "Not Available":np.nan,
    "unknown" : np.nan
    })
```

```python
df[df["Ship Mode"].isnull()]
```

```python
df.isnull().sum()
```

```python
## fix column name
```

```python
df.rename(columns={
    "Order Id":"order_id",
    "Order Date" :"order_date",
    "Ship Mode" : "ship_mode",
    "Postal Code" : "postal_code",
    "Sub Category" : "sub_category",
    "Product Id" : "product_id",
    "cost price" : "cost_price",
    "List Price" : "list_price",
    "Discount Percent" : "discount_perc",
    "Segment" : "segment",
    "Country" : "country",
    "City" : "city",
    "State" : "state"
},inplace=True)
```

```python
df.info()
```

```python
df.columns = df.columns.str.lower()
```

```python
df.columns.str.replace(" ","_")
```

```python
df.head(5)
```

```python
df.info()
```

```python
df["order_date"] = pd.to_datetime(df["order_date"])
```

```python
df["discount_value"] = df["list_price"] * df["discount_perc"]*.01
```

```python
df["sale"] = df["list_price"] - df["discount_value"]
```

```python
df["profit"] = df["sale"] - df["cost_price"]
```

```python
df.head(5)
```

```python
## drop unnecessaru columns
```

```python
df.drop(["cost_price","list_price","discount_perc"],axis=1,inplace=True)
```
